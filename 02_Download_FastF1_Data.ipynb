```python
import fastf1

# Load the 2024 Bahrain Grand Prix Race session
session = fastf1.get_session(2024, "Bahrain", "R")

print("Loading session...")
session.load()

print("Session loaded successfully!")
```

```python
session.results.to_csv("../Data/raw/fastf1/bahrain_2024_results.csv", index=False)

session.laps.to_csv("../Data/raw/fastf1/bahrain_2024_laps.csv", index=False)

print("Files saved successfully!")
```

```python
session.laps.to_csv(
    "../Data/raw/fastf1/bahrain_2024_laps.csv",
    index=False
)

print("Lap data saved!")
```

```python
weather = session.weather_data

weather.head()
```
