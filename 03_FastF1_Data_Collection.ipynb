# Download 2020 Season

```python
import fastf1
import pandas as pd
import os

# Enable cache
fastf1.Cache.enable_cache("../Data/raw/fastf1")

# Create folders
os.makedirs("../Data/raw/fastf1/results", exist_ok=True)
os.makedirs("../Data/raw/fastf1/laps", exist_ok=True)
os.makedirs("../Data/raw/fastf1/weather", exist_ok=True)

# Get the 2020 schedule
schedule = fastf1.get_event_schedule(2020)

# Download every 2020 race
for _, event in schedule.iterrows():

    event_name = event["EventName"]

    print(f"Downloading {event_name}...")

    session = fastf1.get_session(2020, event_name, "R")
    session.load()

    # Save race results
    session.results.to_csv(
        f"../Data/raw/fastf1/results/2020_{event_name}_results.csv",
        index=False
    )

    # Save lap data
    session.laps.to_csv(
        f"../Data/raw/fastf1/laps/2020_{event_name}_laps.csv",
        index=False
    )

    # Save weather data
    session.weather_data.to_csv(
        f"../Data/raw/fastf1/weather/2020_{event_name}_weather.csv",
        index=False
    )

print("2020 season downloaded successfully!")
```

# Download 2021 Season

```python
from pathlib import Path

schedule = fastf1.get_event_schedule(2021)

for _, event in schedule.iterrows():

    event_name = event["EventName"]

    results_file = Path(f"../Data/raw/fastf1/results/2021_{event_name}_results.csv")

    # Skip races that are already downloaded
    if results_file.exists():
        print(f"✓ Skipping {event_name} (already downloaded)")
        continue

    print(f"Downloading {event_name}...")

    session = fastf1.get_session(2021, event_name, "R")
    session.load()

    session.results.to_csv(
        f"../Data/raw/fastf1/results/2021_{event_name}_results.csv",
        index=False
    )

    session.laps.to_csv(
        f"../Data/raw/fastf1/laps/2021_{event_name}_laps.csv",
        index=False
    )

    session.weather_data.to_csv(
        f"../Data/raw/fastf1/weather/2021_{event_name}_weather.csv",
        index=False
    )

print("2021 season downloaded successfully!")
```
