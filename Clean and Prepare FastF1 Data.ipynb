```python
import pandas as pd
from pathlib import Path
import glob
results_folder = "../Data/raw/fastf1/results"
laps_folder = "../Data/raw/fastf1/laps"
weather_folder = "../Data/raw/fastf1/weather"
```

```python
result_files = glob.glob(f"{results_folder}/*.csv")

results_df = pd.concat(
    [pd.read_csv(file) for file in result_files],
    ignore_index=True
)

print(f"Loaded {len(result_files)} results files")
print(results_df.shape)

results_df.head()
```

```python
lap_files = glob.glob(f"{laps_folder}/*.csv")

laps_df = pd.concat(
    [pd.read_csv(file) for file in lap_files],
    ignore_index=True
)

print(f"Loaded {len(lap_files)} lap files")
print(laps_df.shape)

laps_df.head()
```

```python
weather_files = glob.glob(f"{weather_folder}/*.csv")

weather_df = pd.concat(
    [pd.read_csv(file) for file in weather_files],
    ignore_index=True
)

print(f"Loaded {len(weather_files)} weather files")
print(weather_df.shape)

weather_df.head()
```
